```python
dados02 = pd.read_excel(r'C:\Users\Casa\Documents\0.codigos\GitHub\etl\dados\dados02.xlsx')

print(dados02.to_string())
```

```python
# Identifique as linhas onde a coluna 'BAIRRO' NÃO é do tipo string.
condicao_nao_string = dados02['BAIRRO'].apply(lambda x: not isinstance(x, str))

# 2. Substitua em 'LOGRADOURO, NUMERO' para as linhas onde 'BAIRRO' não é string.
#    Aqui, vou adicionar um prefixo para indicar a modificação. Você pode ajustar conforme necessário.
dados02.loc[condicao_nao_string, 'LOGRADOURO, NUMERO'] = "MODIFICADO: " + dados02['LOGRADOURO, NUMERO'].astype(str)

# 3. Defina 'BAIRRO' para um valor que represente "nulo" ou "vazio" seguindo as melhores práticas.
#    A melhor prática para representar valores ausentes no pandas é usar np.nan (Not a Number).
#    Se você preferir uma string vazia, pode usar "".

# Usando np.nan (recomendado):
dados02.loc[condicao_nao_string, 'BAIRRO'] = np.nan

# Ou, se você realmente preferir uma string vazia:
# dados02.loc[condicao_nao_string, 'BAIRRO'] = ""

# Para visualizar as alterações (algumas linhas):
print(dados02[['LOGRADOURO, NUMERO', 'BAIRRO']].head(10))

# Para verificar os tipos após a modificação:
print("\nTipos de dados após a modificação:")
print(dados02.dtypes)
```
